```python
seasonal_performance = df.groupby("Season")["Production"].mean()

print(seasonal_performance)
```

```python
plt.figure(figsize=(8,5))

seasonal_performance.plot(kind="bar")

plt.title("Average Agricultural Performance Across Seasons")
plt.xlabel("Season")
plt.ylabel("Average Production")
plt.xticks(rotation=0)

plt.tight_layout()
plt.show()
```
